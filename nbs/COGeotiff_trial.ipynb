```python
from osgeo import gdal
data = gdal.Open('/vsicurl/http://mobileapp.nesdr.gov.in/out2019.tif')
```

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
out=gdal.Translate('',data, format='MEM', strict=True,width=256, height=256)
vis = np.stack((out.ReadAsArray()[7,...]/4000,out.ReadAsArray()[4,...]/2500,out.ReadAsArray()[3,...]/2500),axis=2)
```
